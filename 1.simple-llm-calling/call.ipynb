```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI


load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it")

response = llm.invoke("How many moons does jupiter have?")
print(response.text)
```

```python
response = llm.invoke([["system","You are a helpful assistant that answers in short one line "],
                ["human", "what will be the future of tech developers because of the development of agi "]])
print(response.text)
```

```python
prompt = "write a single tagline for quirky coffee shop. only output the tagline, nothing else."


llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0)

response = llm.invoke(prompt)
print(response.text)
```

```python

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=2)

response = llm.invoke(prompt)
print(response.text)
```

## Call LLM using GROQ

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="qwen/qwen3.8-27b", reasoning_format="parsed")

message = [
    ("system", "you are a helpful assistant that give a short answer to a question"),
    ("human", "how to be a best ai engineer")
]

ai_msg = llm.invoke(message)
print(ai_msg.content)
```
